# 高维协变量调整与稳健资产配置实验分析

本 notebook 读取蒙特卡洛实验结果，进行交互式探索。

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../output/tables/results.csv')
df.head()
```

## 1. 按估计量聚合的平均表现

```python
summary = df.groupby('estimator')[['bias', 'rmse', 'coverage', 'mean_weight_abs_err', 'mean_utility_loss']].mean()
summary.round(4)
```

## 2. 覆盖率随样本量变化

```python
plt.figure(figsize=(10, 6))
sub = df[df['p'] == df['p'].mode()[0]]
for est in sub['estimator'].unique():
    d = sub[sub['estimator'] == est].sort_values('n')
    plt.plot(d['n'], d['coverage'], marker='o', label=est)
plt.axhline(0.95, color='gray', linestyle='--')
plt.xscale('log')
plt.xlabel('n')
plt.ylabel('Coverage')
plt.legend()
plt.title('Coverage vs Sample Size')
plt.show()
```

## 3. RMSE 热力图

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
axes = axes.flatten()
for ax, est in zip(axes, df['estimator'].unique()):
    sub = df[df['estimator'] == est]
    pivot = sub.pivot_table(values='rmse', index='n', columns='p', aggfunc='mean')
    sns.heatmap(pivot, annot=True, fmt='.3f', cmap='YlOrRd', ax=ax)
    ax.set_title(est)
plt.tight_layout()
plt.show()
```

## 4. 资产配置权重误差

```python
plt.figure(figsize=(10, 6))
sub = df[df['p'] == df['p'].mode()[0]].sort_values('n')
for est in sub['estimator'].unique():
    d = sub[sub['estimator'] == est]
    plt.plot(d['n'], d['mean_weight_abs_err'], marker='o', label=est)
plt.xscale('log')
plt.xlabel('n')
plt.ylabel('Mean absolute weight error')
plt.legend()
plt.title('Portfolio Weight Error vs Sample Size')
plt.show()
```
